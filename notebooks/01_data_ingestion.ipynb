```python
from pathlib import Path
import pandas as pd

# Find the project folder from the current notebook location
candidates = [Path.cwd(), *Path.cwd().parents]

PROJECT_ROOT = next(
    path for path in candidates
    if (path / "data" / "raw" / "01_fund_master.csv").exists()
)

RAW_DIR = PROJECT_ROOT / "data" / "raw"

print("Project root:", PROJECT_ROOT)
print("Raw data folder:", RAW_DIR)
```

```python
csv_files = sorted(RAW_DIR.glob("01_*.csv")) + sorted(RAW_DIR.glob("02_*.csv"))
csv_files += sorted(RAW_DIR.glob("03_*.csv"))
csv_files += sorted(RAW_DIR.glob("04_*.csv"))
csv_files += sorted(RAW_DIR.glob("05_*.csv"))
csv_files += sorted(RAW_DIR.glob("06_*.csv"))
csv_files += sorted(RAW_DIR.glob("07_*.csv"))
csv_files += sorted(RAW_DIR.glob("08_*.csv"))
csv_files += sorted(RAW_DIR.glob("09_*.csv"))
csv_files += sorted(RAW_DIR.glob("10_*.csv"))

csv_files = sorted(set(csv_files))

print(f"Datasets found: {len(csv_files)}")
display([file.name for file in csv_files])
```

```python
datasets = {}
quality_summary = []

for file_path in csv_files:
    df = pd.read_csv(file_path)
    datasets[file_path.name] = df

    print("=" * 80)
    print(f"FILE: {file_path.name}")
    print("Shape:", df.shape)
    print("\nData types:")
    display(df.dtypes.to_frame("dtype"))
    print("\nFirst five rows:")
    display(df.head())

    anomalies = []

    missing_values = df.isna().sum()
    missing_values = missing_values[missing_values > 0]

    if not missing_values.empty:
        anomalies.append(
            "Missing values: "
            + ", ".join(
                f"{column}={count}"
                for column, count in missing_values.items()
            )
        )

    duplicate_rows = df.duplicated().sum()

    if duplicate_rows > 0:
        anomalies.append(f"Duplicate rows: {duplicate_rows}")

    if df.empty:
        anomalies.append("Empty dataset")

    if df.columns.duplicated().any():
        anomalies.append("Duplicate column names")

    if not anomalies:
        anomalies.append("None detected")

    print("\nAnomalies:")
    print("; ".join(anomalies))

    quality_summary.append({
        "file": file_path.name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "anomalies": "; ".join(anomalies)
    })
```

```python
quality_df = pd.DataFrame(quality_summary)
display(quality_df)
```

```python
fund_master = datasets["01_fund_master.csv"]

print("Fund master shape:", fund_master.shape)
display(fund_master.head())
display(fund_master.dtypes.to_frame("dtype"))
print("Unique fund houses:")
display(sorted(fund_master["fund_house"].dropna().unique()))

print("\nUnique categories:")
display(sorted(fund_master["category"].dropna().unique()))

print("\nUnique sub-categories:")
display(sorted(fund_master["sub_category"].dropna().unique()))

print("\nUnique risk categories:")
display(sorted(fund_master["risk_category"].dropna().unique()))

print("Unique fund houses:")
display(sorted(fund_master["fund_house"].dropna().unique()))

print("\nUnique categories:")
display(sorted(fund_master["category"].dropna().unique()))

print("\nUnique sub-categories:")
display(sorted(fund_master["sub_category"].dropna().unique()))

print("\nUnique risk categories:")
display(sorted(fund_master["risk_category"].dropna().unique()))

print("AMFI code data type:")
print(fund_master["amfi_code"].dtype)

print("\nNumber of unique AMFI codes:")
print(fund_master["amfi_code"].nunique())

print("\nSample AMFI codes:")
display(fund_master[["amfi_code", "scheme_name"]].head(10))
```

```python
nav_history = datasets["02_nav_history.csv"]

fund_master_codes = set(
    fund_master["amfi_code"].dropna().astype(str).str.strip()
)

nav_history_codes = set(
    nav_history["amfi_code"].dropna().astype(str).str.strip()
)

missing_codes = sorted(fund_master_codes - nav_history_codes)

print("AMFI codes in fund_master:", len(fund_master_codes))
print("AMFI codes in nav_history:", len(nav_history_codes))
print("Codes missing from nav_history:", len(missing_codes))
print("Missing codes:", missing_codes)
```

```python
print("NAV history shape:", nav_history.shape)
display(nav_history.head())

print("Date range:")
print(nav_history["date"].min(), "to", nav_history["date"].max())

print("\nNAV records by AMFI code:")
display(
    nav_history.groupby("amfi_code")
    .size()
    .reset_index(name="record_count")
    .sort_values("record_count", ascending=False)
)
```

```python
REPORT_DIR = PROJECT_ROOT / "reports"
REPORT_DIR.mkdir(exist_ok=True)

notebook_summary = pd.DataFrame({
    "metric": [
        "Datasets loaded",
        "Fund-master rows",
        "NAV-history rows",
        "Unique fund houses",
        "Unique categories",
        "Unique sub-categories",
        "Unique risk categories",
        "Fund-master AMFI codes",
        "Missing AMFI codes"
    ],
    "value": [
        len(datasets),
        len(fund_master),
        len(nav_history),
        fund_master["fund_house"].nunique(),
        fund_master["category"].nunique(),
        fund_master["sub_category"].nunique(),
        fund_master["risk_category"].nunique(),
        len(fund_master_codes),
        len(missing_codes)
    ]
})

display(notebook_summary)

notebook_summary.to_csv(
    REPORT_DIR / "day1_eda_summary.csv",
    index=False
)

print("Saved:", REPORT_DIR / "day1_eda_summary.csv")
```
